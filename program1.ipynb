```python
#program to merge 2 python dictionary
# Create first dictionary
name1 = ("Blessy", "Tina")
values1 = (18, 20)
dict1 = dict(zip(name1, values1))

# Create second dictionary
name2 = ("Cinu", "Diya")
values2 = (20, 21)
dict2 = dict(zip(name2, values2))

# Merge the dictionaries
merged = {**dict1, **dict2}

print("Dictionary 1:", dict1)
print("Dictionary 2:", dict2)
print("Merged Dictionary:", merged)
```

```python
#program to write a function that takes 2 lists and return true if they have at least one common member
def common(list1,list2):
    for item in list1:
        if item in list2:
            return True
    return False

list1=list(map(int,input("enter elements separated by space:").split()))
list2=list(map(int,input("enter elements separated by space:").split()))

print(common(list1,list2))
```
